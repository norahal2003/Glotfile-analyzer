```python

from google.colab import files
import zipfile
import os

# Upload ADFA-LD ZIP
uploaded = files.upload()

# Extract ZIP
zip_path = next(iter(uploaded.keys()))

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall('/content/ADFA-LD')

print("Dataset extracted successfully!")

# Display folders
for root, dirs, filenames in os.walk('/content/ADFA-LD'):
    print(root, "Files:", len(filenames))
```

```python

import os
from pathlib import Path
import pandas as pd
import numpy as np

# Dataset location
DATASET_PATH = Path("/content/ADFA-LD")

# Find all TXT files
all_files = list(DATASET_PATH.rglob("*.txt"))

print("Total TXT files:", len(all_files))

# Count files by category
categories = {
    "Training": [],
    "Validation": [],
    "Attack": []
}

for file in all_files:
    path = str(file).lower()

    if "training_data_master" in path:
        categories["Training"].append(file)

    elif "validation_data_master" in path:
        categories["Validation"].append(file)

    elif "attack_data_master" in path:
        categories["Attack"].append(file)

# Display counts
print("\nDataset Summary")

for category, files in categories.items():
    print(f"{category}: {len(files)} files")

# Display attack types
print("\nAttack Types:")

attack_folder = DATASET_PATH / "Attack_Data_Master"

if attack_folder.exists():
    for folder in sorted(attack_folder.iterdir()):
        if folder.is_dir():
            count = len(list(folder.rglob("*.txt")))
            print(folder.name, ":", count, "files")
```

```python

records = []

for category, files in categories.items():

    for file in files:

        try:
            content = file.read_text().strip()
            values = content.split()

            records.append({
                "file": file.name,
                "category": category,
                "sequence_length": len(values),
                "empty": len(values) == 0,
                "valid": all(v.isdigit() for v in values)
            })

        except Exception as e:
            print("Error reading:", file, e)

df = pd.DataFrame(records)

# Display first records
print(df.head(10))

# Sequence statistics
print("\nSequence Length Statistics:")
print(df.groupby("category")["sequence_length"].describe())

# Check empty files
print("\nEmpty files:", df["empty"].sum())

# Check invalid files
print("Invalid files:", (~df["valid"]).sum())
```

```python

import matplotlib.pyplot as plt

df.boxplot(
    column="sequence_length",
    by="category",
    figsize=(8, 5)
)

plt.title("System Call Sequence Length Distribution")
plt.suptitle("")
plt.xlabel("Dataset Category")
plt.ylabel("Number of System Calls")
plt.show()
```

```python

from pathlib import Path
import pandas as pd

BASE = Path("/content/ADFA-LD/ADFA-LD")

records = []

# Normal training records
for file in (BASE / "Training_Data_Master").rglob("*.txt"):
    sequence = file.read_text().strip()

    if sequence:
        records.append({
            "sequence": sequence,
            "label": 0,
            "attack_type": "Normal",
            "file_name": file.name
        })

# Attack records
for file in (BASE / "Attack_Data_Master").rglob("*.txt"):
    sequence = file.read_text().strip()

    if sequence:
        attack_type = file.parent.name.rsplit("_", 1)[0]

        records.append({
            "sequence": sequence,
            "label": 1,
            "attack_type": attack_type,
            "file_name": file.name
        })

# Create dataframe
dataset = pd.DataFrame(records)

print("Dataset shape:", dataset.shape)
print("\nLabel distribution:")
print(dataset["label"].value_counts())

print("\nAttack distribution:")
print(dataset["attack_type"].value_counts())

print("\nFirst records:")
display(dataset.head())
```
